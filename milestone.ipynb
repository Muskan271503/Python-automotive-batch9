# Q1

Problem Statement

Arjun is a budding stock market enthusiast trying to analyze past stock prices to make smarter trading decisions. He has come up with two metrics to evaluate a given sequence of stock prices:

1. Max Profit: This is the highest possible profit he could earn by buying at one point and selling later.
2. Volatility Index: This metric reflects the average day-to-day fluctuation in prices.

Given a list of stock prices over several days, help Arjun compute both these values to inform his future investment decisions.

Complete the class StockAnalyzer to perform the necessary computations.

Example:
Input:
1 2 3 4 5
Output:
Max Profit: 4
Volatility Index: 1.00

Explanation:
Step 1: Max Profit Calculation
The prices are: 1, 2, 3, 4, 5
The best scenario is buying at 1 (the lowest price) and selling at 5 (the highest price after 1).
So, the max profit is: 5 - 1 = 4

Step 2: Volatility Index Calculation
The day-to-day absolute differences are:
|2 - 1| = 1
|3 - 2| = 1
|4 - 3| = 1
|5 - 4| = 1
Sum of these differences = 1 + 1 + 1 + 1 = 4
Average (volatility index) = 4 / 4 = 1.00

So the final output is:
Max Profit: 4
Volatility Index: 1.00


```python
class StockAnalyzer:
    def __init__(self, price_data):
        self.price_data = price_data

    def find_best_profit(self):
        lowest_rate = self.price_data[0]
        highest_gain = 0

        for current_rate in self.price_data[1:]:
            if current_rate < lowest_rate:
                lowest_rate = current_rate
            else:
                profit_today = current_rate - lowest_rate
                if profit_today > highest_gain:
                    highest_gain = profit_today

        return highest_gain

    def calculate_volatility(self):
        fluctuation_total = 0
        total_days = len(self.price_data) - 1

        for i in range(1, len(self.price_data)):
            daily_change = abs(self.price_data[i] - self.price_data[i - 1])
            fluctuation_total += daily_change

        return fluctuation_total / total_days


# -------- Input Section --------
prices = list(map(int, input().split()))

# -------- Object Creation --------
analysis = StockAnalyzer(prices)

# -------- Output Section --------
max_gain = analysis.find_best_profit()
volatility_score = analysis.calculate_volatility()

print(f"Max Profit: {max_gain}")
print(f"Volatility Index: {volatility_score:.2f}")
```

```python
class StockAnalyzer:
    def __init__(self):
        self.daily_rates = [1, 2, 3, 4, 5]   # predefined stock prices

    def best_possible_profit(self):
        base_price = self.daily_rates[0]
        max_earning = 0

        for rate in self.daily_rates[1:]:
            if rate < base_price:
                base_price = rate
            else:
                earning = rate - base_price
                if earning > max_earning:
                    max_earning = earning

        return max_earning

    def volatility_measure(self):
        total_variation = 0
        count = len(self.daily_rates) - 1

        for index in range(count):
            total_variation += abs(
                self.daily_rates[index + 1] - self.daily_rates[index]
            )

        return total_variation / count


# -------- Program Execution --------
stock_report = StockAnalyzer()

profit_result = stock_report.best_possible_profit()
volatility_result = stock_report.volatility_measure()

print("Max Profit:", profit_result)
print(f"Volatility Index: {volatility_result:.2f}")
```

Problem Statement

Write a program that calculates the average of a list of integers. The program first prompts the user to enter the length of the list (n), followed by each element of the list. The program should include error handling to ensure that:
The length of the list (n) is a positive integer.
Each input element is a valid integer.

If the user enters invalid data at any point, the program should terminate with the appropriate error message. The error handling should cover the following scenarios:

If n is not a valid positive integer, print:
Error: The length of the list must be a non-negative integer.

If a non-numeric value is entered for n, print:
Error: You must enter a numeric value.

If any of the elements in the list are non-numeric, print:
Error: You must enter a numeric value.

If all inputs are valid, the program should calculate the average of the list of integers and print it with two decimal places.


```python
try:
    size_input = input()
    
    # check if n is numeric
    if not size_input.isdigit():
        print("Error: You must enter a numeric value.")
        exit()

    total_numbers = int(size_input)

    # check if n is positive
    if total_numbers < 0:
        print("Error: The length of the list must be a non-negative integer.")
        exit()

    values = []

    for _ in range(total_numbers):
        element = input()

        # check if list element is numeric (handles negatives too)
        if element.lstrip('-').isdigit():
            values.append(int(element))
        else:
            print("Error: You must enter a numeric value.")
            exit()

    # calculate average
    if total_numbers > 0:
        average_result = sum(values) / total_numbers
        print(f"{average_result:.2f}")
    else:
        print("0.00")

except:
    print("Error: You must enter a numeric value.")
```

```python
class AverageCalculator:
    def __init__(self):
        # predefined list length and elements
        self.count = 5
        self.numbers = [10, 20, 30, 40, 50]

    def validate_and_compute(self):
        # validate length
        if not isinstance(self.count, int):
            print("Error: You must enter a numeric value.")
            return

        if self.count < 0:
            print("Error: The length of the list must be a non-negative integer.")
            return

        # validate elements
        for value in self.numbers:
            if not isinstance(value, int):
                print("Error: You must enter a numeric value.")
                return

        # calculate average
        if self.count == 0:
            print("0.00")
        else:
            avg = sum(self.numbers) / self.count
            print(f"{avg:.2f}")


# -------- Program Run --------
avg_finder = AverageCalculator()
avg_finder.validate_and_compute()
```

Problem Statement

Mary is managing a business and wants to analyze its profitability. She operates both a regular business model and a seasonal business model. To assess profitability, she uses a program that calculates and compares the profit margins for both models based on revenue and cost.

The program defines the following:

BusinessUtility class: This class contains a method calculate_margin(double revenue, double cost) to compute the regular profit margin.
SeasonalBusinessUtility class: This class inherits from BusinessUtility and overrides the calculate_margin(double revenue, double cost) method. It adds a seasonal adjustment of 10% to the base margin to compute the seasonal margin.
ProfitabilityChecker class: This class contains the method check_profitability(double regular_margin), which checks whether the business is profitable based on the regular margin. If the regular margin is 10% or higher, it prints "Business is profitable." Otherwise, it prints "Business is not profitable."

Mary inputs revenue and cost, and the program computes and display the regular and seasonal margins using:
Regular Margin = ((Revenue − Cost) / Revenue) × 100
Seasonal Margin = Regular Margin + 10


```python
# Base class for normal business calculations
class BusinessUtility:
    def calculate_margin(self, sales_amount, expense_amount):
        margin = ((sales_amount - expense_amount) / sales_amount) * 100
        return margin


# Derived class for seasonal business calculations
class SeasonalBusinessUtility(BusinessUtility):
    def calculate_margin(self, sales_amount, expense_amount):
        base_margin = super().calculate_margin(sales_amount, expense_amount)
        seasonal_bonus = 10
        return base_margin + seasonal_bonus


# Class to check profitability condition
class ProfitabilityChecker:
    def check_profitability(self, margin_value):
        if margin_value >= 10:
            print("Business is profitable.")
        else:
            print("Business is not profitable.")


# -------- Hard-coded values --------
revenue_value = 50000
cost_value = 42000

# -------- Object creation --------
regular_business = BusinessUtility()
seasonal_business = SeasonalBusinessUtility()
profit_check = ProfitabilityChecker()

# -------- Calculations --------
regular_margin = regular_business.calculate_margin(revenue_value, cost_value)
seasonal_margin = seasonal_business.calculate_margin(revenue_value, cost_value)

# -------- Output --------
print(f"Regular Margin: {regular_margin:.2f}%")
print(f"Seasonal Margin: {seasonal_margin:.2f}%")

profit_check.check_profitability(regular_margin)
```

```python
# Base class for regular business margin calculation
class BusinessUtility:
    def calculate_margin(self, income, spending):
        return ((income - spending) / income) * 100


# Derived class with seasonal adjustment
class SeasonalBusinessUtility(BusinessUtility):
    def calculate_margin(self, income, spending):
        normal_margin = super().calculate_margin(income, spending)
        seasonal_extra = 10
        return normal_margin + seasonal_extra


# Class to verify profitability
class ProfitabilityChecker:
    def check_profitability(self, margin_score):
        if margin_score >= 10:
            print("Business is profitable.")
        else:
            print("Business is not profitable.")


# -------- Input Section --------
revenue = float(input("Enter revenue: "))
cost = float(input("Enter cost: "))

# -------- Object Creation --------
basic_business = BusinessUtility()
seasonal_business = SeasonalBusinessUtility()
profit_status = ProfitabilityChecker()

# -------- Calculations --------
regular_margin = basic_business.calculate_margin(revenue, cost)
seasonal_margin = seasonal_business.calculate_margin(revenue, cost)

# -------- Output --------
print(f"Regular Margin: {regular_margin:.2f}%")
print(f"Seasonal Margin: {seasonal_margin:.2f}%")

profit_status.check_profitability(regular_margin)
```
