```python
# decorator
def changecase(func):
    def myinner():
        return func().upper()
    return myinner


@changecase
def myfunc():
    return "hi"


print(myfunc())

```

```python
def myWrapper(func):
    def myInnerFunc():
        print("Inside wrapper.")
        func()
    return myInnerFunc

@myWrapper
def myFunc():
    print("Hello, World!")
    
myFunc()
```

```python
def log_function(func):
    def logger(*args, **kwargs):
        print(">>> Starting:", func.__name__)
        output = func(*args, **kwargs)
        print("<<< Ending:", func.__name__)
        return output
    return logger
@log_function
def greet(name):
    print(f"Hello, {name}!")

greet("Muskan")
```

```python
def log_function(func):
    def wrapper(*args, **args2):
        print(f"Calling function: {func.__name__}")
        result = func(*args, **args2)
        print(f"function name: {func.__name__}")
        return result
    return wrapper


@log_function
def greeting(name):
    return f"hello, {name}"


#Function call added
print(greeting("Muskan"))
```

```python
def calculator(func):
    def wrapper(a, b):
        print("Operation:", func.__name__)
        result = func(a, b)
        print("Result:", result)
        print("-----------------")
        return result
    return wrapper


@calculator
def add(a, b):
    return a + b


@calculator
def subtract(a, b):
    return a - b


# FUNCTION CALLS (important)
add(20, 10)
subtract(20, 10)
```

```python
def star_decorator(func):
    def wrapper():
        return "*****" + func() + "*****"
    return wrapper

@star_decorator
def say_hello():
    return "Hello"

print(say_hello())

```

```python
def calculator(func):
    def wrapper(a, b):
        print("\nNumbers entered:", a, "and", b)
        print("Operation:", func.__name__)
        result = func(a, b)
        print(f"Result of {func.__name__}({a}, {b}) = {result}")
        return result
    return wrapper

@calculator
def add(a, b):
    return a + b
@calculator
def subtract(a, b):
    return a - b

print("CALCULATOR USING DECORATOR")
print("1. Addition")
print("2. Subtraction")

choice = int(input("Enter your choice (1 or 2): "))

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

if choice == 1:
    add(num1, num2)
elif choice == 2:
    subtract(num1, num2)
else:
    print("Invalid choice")
```

```python
def calculator(func):
    def wrapper():
        # Ask user for operation
        choice = input("Do you want to Add or Subtract? (add/sub): ").lower()
        if choice not in ["add", "sub"]:
            print("Invalid choice!")
            return

        # Take numbers from user
        try:
            a = float(input("Enter first number: "))
            b = float(input("Enter second number: "))
        except ValueError:
            print("Please enter valid numbers!")
            return

        # Call the decorated function
        result = func(a, b, choice)
        print(f"\nYou chose to {choice} {a} and {b}.")
        print(f"Result: {result}\n")
    return wrapper


@calculator
def math_operation(a, b, operation):
    if operation == "add":
        return a + b
    else:  # subtract
        return a - b


print("WELCOME TO DECORATOR CALCULATOR\n")
math_operation()
```

```python
def calculator(func):
    def inner(a, b):
        return func(a, b)
    return inner

@calculator
def add(x, y):
    return x + y

@calculator
def sub(x, y):
    return x - y    

@calculator
def multiply(x, y):
    return x * y

@calculator
def divide(x, y):
    if y != 0:
        return x / y
    else:
        return "Error: Division by zero"


print("Select operation:")
print("add, sub, multiply, divide")
choice = input("Enter choice of operation: ")


# Convert inputs to floats to handle decimals
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

if choice == 'add':
    print(f"Result: {add(num1, num2)}")
elif choice == 'sub':
    print(f"Result: {sub(num1, num2)}")
elif choice == 'multiply':
    print(f"Result: {multiply(num1, num2)}")
elif choice == 'divide':
    print(f"Result: {divide(num1, num2)}")
else:
    print("Invalid Input")
```
