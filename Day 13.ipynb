```python
#lambda function - anonymous function that takes any number of arguments, but can only have one expression
x = lambda a : a + 10
print(x(5))

y = lambda a, b : a * b
print(y(2, 3))

z = lambda a, b, c : a + b + c
print(z(1, 2, 3))

```

```python
def count(n):
    if n <= 5:
        print("a")
    else:
        print(n)
        count(n - 1)
        
count(5)

#recursion - when a function calls itself
```

```python
# generatior are memory efficient because they generate items one by one and only when requested, rather than storing all items in memory at once.
def abc():
    yield 1 # yield keyword make a def/fun to convert into generator
    yield 2
    yield 3 
    
for value in abc():
    print(value)
```

```python
def count(n):
    c = 1
    while c <= n:
        yield c
        c += 1
        
for num in count(4):
    print(num)
```

```python
def large_num(n):
    for i in range(n):
            yield i
#this doesn't create a lakh numbers in memory at once, it generates one number at a time when requested.           
gen = large_num(1000000)
print(next(gen))
print(next(gen))
print(next(gen))

```

```python
# list comprehension creates the entire list in memory at once.
list = [x*x for x in range(5)]
print(list)

#doing comprehension using generator expression
gen = (x*x for x in range(5))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
# dictionary comprehension
num_dict = {num: num * 2 for num in range(1, 4)}
print(num_dict)


states = ["AP", "WB", "TN"]
capitals = ["Amaravati", "Kolkata", "Chennai"]

state_capital = {state: capital for state, capital in zip(states, capitals)}
print(state_capital)

# set comprehension
a = [1, 1, 2, 2]
s = {n for n in a if n % 2 == 0}
print(s)
```

```python
from functools import reduce
numbers = [1, 2, 3, 4, 5]

# The lambda x, y: x * y multiplies the current accumulator (x) and the next item (y)
product_result = reduce(lambda x, y: x * y, numbers)
print(product_result)
```

```python
from functools import reduce

# -------------------------------
# Step 1: Define Managers
# -------------------------------
managers = ["A", "B", "C"]

# -------------------------------
# Step 2: Define Employees (12)
# Each employee has only ONE manager
# -------------------------------
employees = [
    {"id": 1, "name": "Emp1", "manager": "A", "performance": 85},
    {"id": 2, "name": "Emp2", "manager": "A", "performance": 72},
    {"id": 3, "name": "Emp3", "manager": "A", "performance": 90},
    {"id": 4, "name": "Emp4", "manager": "A", "performance": 60},

    {"id": 5, "name": "Emp5", "manager": "B", "performance": 88},
    {"id": 6, "name": "Emp6", "manager": "B", "performance": 70},
    {"id": 7, "name": "Emp7", "manager": "B", "performance": 95},
    {"id": 8, "name": "Emp8", "manager": "B", "performance": 55},

    {"id": 9, "name": "Emp9", "manager": "C", "performance": 92},
    {"id": 10, "name": "Emp10", "manager": "C", "performance": 65},
    {"id": 11, "name": "Emp11", "manager": "C", "performance": 78},
    {"id": 12, "name": "Emp12", "manager": "C", "performance": 81},
]

# -------------------------------
# Step 3: Manager → Employees Mapping
# (Dictionary Comprehension)
# -------------------------------
manager_employee_map = {
    manager: [emp["name"] for emp in employees if emp["manager"] == manager]
    for manager in managers
}

print("Manager - Employee Mapping:")
for m, e in manager_employee_map.items():
    print(m, ":", e)

# -------------------------------
# Step 4: High Performers
# Using filter() + lambda
# -------------------------------
high_performers = list(
    filter(lambda emp: emp["performance"] >= 80, employees)
)

print("\nHigh Performing Employees (>=80):")
for emp in high_performers:
    print(emp["name"], "-", emp["performance"])

# -------------------------------
# Step 5: High Performers per Manager
# (Comprehension)
# -------------------------------
manager_high_performers = {
    manager: [
        emp["name"]
        for emp in employees
        if emp["manager"] == manager and emp["performance"] >= 80
    ]
    for manager in managers
}

print("\nHigh Performers per Manager:")
for m, e in manager_high_performers.items():
    print(m, ":", e)

# -------------------------------
# Step 6: Average Performance per Manager
# (filter + reduce + lambda)
# -------------------------------
print("\nAverage Performance per Manager:")

for manager in managers:
    manager_emps = list(
        filter(lambda emp: emp["manager"] == manager, employees)
    )

    total_score = reduce(
        lambda x, y: x + y,
        [emp["performance"] for emp in manager_emps]
    )

    average = total_score / len(manager_emps)
    print(manager, ":", average)
```

```python
from functools import reduce

# -----------------------------------
# Step 1: Input Managers
# -----------------------------------
num_managers = int(input("Enter number of managers: "))
managers = []

for i in range(num_managers):
    manager = input(f"Enter manager {i+1} name: ")
    managers.append(manager)

# -----------------------------------
# Step 2: Input Employees
# -----------------------------------
num_employees = int(input("\nEnter number of employees: "))
employees = []

for i in range(num_employees):
    print(f"\nEnter details for Employee {i+1}")
    name = input("Employee Name: ")
    manager = input("Manager Name: ")
    performance = int(input("Performance Score (0-100): "))

    employees.append({
        "name": name,
        "manager": manager,
        "performance": performance
    })

# -----------------------------------
# Step 3: Manager → Employees Mapping
# (Dictionary Comprehension)
# -----------------------------------
manager_employee_map = {
    manager: [emp["name"] for emp in employees if emp["manager"] == manager]
    for manager in managers
}

print("\nManager - Employee Mapping:")
for m, e in manager_employee_map.items():
    print(m, "->", e)

# -----------------------------------
# Step 4: High Performers (filter + lambda)
# -----------------------------------
high_performers = list(
    filter(lambda e: e["performance"] >= 80, employees)
)

print("\nHigh Performing Employees (>=80):")
for emp in high_performers:
    print(emp["name"], "-", emp["performance"])

# -----------------------------------
# Step 5: Average Performance Per Manager
# (filter + map + reduce)
# -----------------------------------
manager_avg_performance = {}

for manager in managers:
    manager_emps = list(filter(lambda e: e["manager"] == manager, employees))

    if manager_emps:  # avoid division by zero
        scores = list(map(lambda e: e["performance"], manager_emps))
        avg = reduce(lambda x, y: x + y, scores) / len(scores)
        manager_avg_performance[manager] = avg
    else:
        manager_avg_performance[manager] = 0

print("\nAverage Performance Per Manager:")
for m, avg in manager_avg_performance.items():
    print(m, ":", round(avg, 2))

# -----------------------------------
# Step 6: Best Manager Based on Performance
# -----------------------------------
best_manager = max(
    manager_avg_performance.items(),
    key=lambda x: x[1]
)

print("\n🏆 Best Manager Based on Team Performance:")
print("Manager:", best_manager[0])
print("Average Score:", round(best_manager[1], 2))
```
